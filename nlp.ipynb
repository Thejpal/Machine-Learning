## **Tokenization**

Converts paragraphs to sentences or sentences to words. Here, both sentences and words can be tokens

```python
import nltk

nltk.download("punkt_tab")
```

```python
from nltk import word_tokenize, sent_tokenize
```

```python
# Corpus - Also known as the text
# Tokens - Can be either words or sentences

random_para = "It was just a burger. Why couldn't she understand that? She knew he'd completely changed his life around her eating habits, so why couldn't she give him a break this one time? She wasn't even supposed to have found out. Yes, he had promised her and yes, he had broken that promise, but still in his mind, all it had been was just a burger."
```

### **Sentence tokenization**

* Converting a paragraph into a list of sentences

```python
sent_tokenize(random_para)
```

### **Word tokenization**

* Converting a sentence or a paragraph into words

```python
word_tokenize(random_para)
```

## **Stemming**

Stemming is a process of converting a word to it's word stem by removing suffixes or prefixes

```python
from nltk.stem import PorterStemmer, RegexpStemmer, SnowballStemmer
```

```python
porter_stemmer = PorterStemmer()

# Removes the suffixes from the word based on the provided regular expression and $ symbol's position.
# If $ is not provided, it will remove the expression from both the start and end of the word.
# Min parameter specifies the length of the word required for the word to be stemmed.
regexp_stemmer = RegexpStemmer(regexp = "ing$|s$|es$|able$", min = 4)

# An advanced version of the Porter stemmer which also stems the words based on languages
snowball_stemmer = SnowballStemmer(language = "english")
```

```python
words = ["running", "eating", "eats", "advisable", "fairly", "history", "writing", "eaten", "went", "goes", "gone", "sportingly", "congratulations"]
```

```python
stemmed_words = [f"{word} ---> {porter_stemmer.stem(word)}, {regexp_stemmer.stem(word)}, {snowball_stemmer.stem(word)}" for word in words]
```
